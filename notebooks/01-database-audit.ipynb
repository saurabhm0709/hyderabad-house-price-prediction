```python
import numpy as np
import pandas as pd
import sqlite3
```

```python
connection = sqlite3.connect("../data/raw/listings_master.db")
```

Extracting the tables from the database and putting them into CSV files, so that we can work directly with CSV files using pandas

```python
tables = {
    'listings':'listings.csv',
    'locality_ratings': 'locality.csv'
}

for (table,filename) in tables.items():
    query = f"SELECT * FROM {table}"
    output_path = f"../data/extracted/{filename}"
    df = pd.read_sql_query(query,connection)
    df.to_csv(output_path,index=False)
    print(f"Saved {filename} | Shape: {df.shape}")
```

Merging both the CSV files using left join

```python
listings = pd.read_csv("../data/extracted/listings.csv")
locality = pd.read_csv("../data/extracted/locality.csv")
merged_df = pd.merge(
    listings,
    locality,
    on = 'locality_id',
    how = 'left',
    suffixes = ("","_locality")
)

merged_df.to_csv("../data/extracted/merged_listings.csv",index=False)
```
