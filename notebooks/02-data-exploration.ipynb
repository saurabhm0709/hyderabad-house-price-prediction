## Fixing the area unit

```python
area_plain = pd.to_numeric(df['covered_area_sqft'],errors='coerce')
area_raw = pd.to_numeric(df['covered_area_sqft'].astype(str).str.replace(',','',regex=False),errors='coerce')
```

```python
print("readable without stripping commas:", area_plain.notna().sum())
print("readable after stripping commas  :", area_raw.notna().sum())
print("rows rescued                     :", area_raw.notna().sum() - area_plain.notna().sum())
```
